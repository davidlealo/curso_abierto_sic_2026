2.- Aplicar 10% de descuento a una lista

```python
precios = [100, 250, 50, 1200]
for precio in precios:
  descuento = precio * 0.10
  precio_final = precio - descuento
  print(f"Original : ${precio} -> Con descuento: ${precio_final}")
```

3.- Contar letras 'a' en una frase.

```python
frase = "Aprendiendo a programar con Python"
contador = 0
for letra in frase:
  if letra.lower() == 'a':
    contador += 1
print(f"La frase tiene {contador} letras 'a'.")
```

4.- Lista de cuadrados del 1 al 10

```python
cuadrados = []
for n in range (1, 11):
  resultado = n ** 2
  cuadrados.append(resultado)

print(f"Lista de cuadrados: {cuadrados}")
```

5.- Filtrar nombres que empiecen con 'A'.


```python
nombres = ["Ana", "Beto", "Alberto", "Carlos", "Alicia"]
print("Nombres que empiezan con 'A':")
for nombre in nombres:
  if nombre.startswith("A"):
    print(f"- {nombre}")
```

2.- Centiela While msg != "salir"

```python
msg = ""
while msg.lower() != "salir":
  msg = input("Escribe algo (o 'salir' para terminar):")
  print(f"Dijiste: {msg}")
```

3.- Validación While pin "1234".

```python
pin = ""
while pin != "1234":
  pin = input("Introduce tu PIN de seguridad: ")
  if pin != "1234":
    print(f"PIN incorrecto. Intenta de nuevo")

print(f"Acceso concedido")
```

4.- Pop While. lista

```python
tareas = ["Lavar platos", "Estudiar", "Hacer ejercicio"]
while tareas:
  tarea_actual = tareas.pop()
  print(f"Completado: {tarea_actual}")

print(f"Ya no quedan tareas pendientes")
```

5.- Acumulador dinámico

```python
total = 0
numero =- 1
print(f"Calculadora de suma (escribe 0 para ver el total)")
while numero != 0:
  numero = int(input("Ingresa un numero:"))
  total += numero
  print(f"Suma actual: {total}")

print(f"---El total final es: {total}---")
```
