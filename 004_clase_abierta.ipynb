### Naming Conventions Examples

There are several common naming conventions used in programming to improve code readability and maintainability. Here are some examples:

#### 1. camelCase (Lower Camel Case)

*   **Description**: The first letter of the first word is lowercase, and the first letter of each subsequent concatenated word is capitalized.
*   **Common Use**: Variables, functions, and methods in languages like JavaScript, Java, and sometimes Python (though `snake_case` is preferred for variables in Python).

````python
# Example of camelCase
myVariableName = "This is a camel case variable"
calculateTotalPrice = lambda price, quantity: price * quantity

print(myVariableName)
print(calculateTotalPrice(10, 5))
````

#### 2. snake_case

*   **Description**: All letters are lowercase, and words are separated by underscores (`_`).
*   **Common Use**: Variables, functions, and module names in Python (PEP 8 standard), and database column names.

````python
# Example of snake_case
my_variable_name = "This is a snake case variable"
calculate_total_price = lambda price, quantity: price * quantity

print(my_variable_name)
print(calculate_total_price(10, 5))
````

#### 3. PascalCase (Upper Camel Case)

*   **Description**: The first letter of each concatenated word is capitalized.
*   **Common Use**: Class names in many object-oriented languages (including Python, as per PEP 8), and sometimes types or constructors.

````python
# Example of PascalCase
class MyClassName:
    def __init__(self, value):
        self.value = value

def my_function_name():
    pass # Function names should typically be snake_case

my_object = MyClassName(100)
print(f"Created an object of class MyClassName with value: {my_object.value}")
````

#### 4. kebab-case

*   **Description**: All letters are lowercase, and words are separated by hyphens (`-`).
*   **Common Use**: File names, URL slugs, and sometimes CSS class names. It is generally **not used** for variable or function names in most programming languages because the hyphen is interpreted as a subtraction operator.

#### 5. SCREAMING_SNAKE_CASE

*   **Description**: All letters are uppercase, and words are separated by underscores (`_`).
*   **Common Use**: Constants and immutable values.

````python
# Example of SCREAMING_SNAKE_CASE
MAX_VALUE = 1000
DEFAULT_TIMEOUT_SECONDS = 30

print(f"Maximum allowed value: {MAX_VALUE}")
print(f"Default timeout: {DEFAULT_TIMEOUT_SECONDS} seconds")
````

### Rules for Writing Variable Names in Python

In Python, there are specific rules and conventions for naming variables to ensure your code is readable and doesn't lead to syntax errors. Here are the main rules:

#### 1. Cannot start with a number

Variable names cannot begin with a numeric digit. They must start with a letter (a-z, A-Z) or an underscore (`_`).

````python
# Valid:
_my_variable = 10
variable_1 = 20

# Invalid (will raise a SyntaxError if uncommented):
# 1variable = 30
# 2nd_name = 'Python'

print(f"_my_variable: {_my_variable}")
print(f"variable_1: {variable_1}")
````

#### 2. Can only contain alphanumeric characters and underscores

Variable names can only contain letters (a-z, A-Z), numbers (0-9), and underscores (`_`). Special characters like `!`, `@`, `#`, `$`, `%`, etc., are not allowed.

````python
# Valid:
first_name = "Alice"
user_id_007 = 7

# Invalid (will raise a SyntaxError if uncommented):
# user-name = "Bob" # Hyphen is not allowed
# total$ = 100 # Dollar sign is not allowed

print(f"first_name: {first_name}")
print(f"user_id_007: {user_id_007}")
````

#### 3. Case-sensitive

Python variable names are case-sensitive. `myVariable` is different from `myvariable`.

````python
# Example of case-sensitivity
myVar = 5
myvar = 10

print(f"myVar: {myVar}")
print(f"myvar: {myvar}")
````

#### 4. Cannot be Python keywords (reserved words)

You cannot use any of Python's reserved keywords as variable names. Keywords are words that have special meaning in Python, like `if`, `else`, `while`, `for`, `class`, `def`, `import`, `True`, `False`, `None`, etc.

````python
import keyword

# Print all Python keywords
print("Python Keywords:", keyword.kwlist)

# Valid:
my_if_statement = "valid"

# Invalid (will raise a SyntaxError if uncommented):
# class = "MyClass"
# import = "module"

print(f"my_if_statement: {my_if_statement}")
````

#### 5. Short and descriptive

While not a strict rule, it's a strong convention (PEP 8) to use short, descriptive names that indicate the purpose of the variable. Avoid overly generic names like `x` or `data` unless their context is extremely clear. Python prefers `snake_case` for variable and function names.

### Cuadro de Precedencia de Operadores

El **cuadro de precedencia de operadores** (o jerarquía de operadores) es un conjunto de reglas que define el orden en el que se evalúan las operaciones en una expresión matemática o lógica. Al igual que en las matemáticas, donde la multiplicación y división se realizan antes que la suma y la resta, los lenguajes de programación siguen reglas similares para evitar ambigüedades y asegurar que las expresiones se evalúen de manera consistente.

#### ¿Por qué es importante?

La precedencia de operadores es crucial porque determina el resultado de una expresión. Si no se respetan estas reglas, o si no se utilizan paréntesis para forzar un orden específico, el resultado podría ser diferente al esperado.

#### Reglas Generales (de mayor a menor precedencia):

Aunque el orden exacto puede variar ligeramente entre lenguajes y para operadores muy específicos, las reglas generales suelen ser las siguientes:

1.  **Paréntesis `()`**: Las operaciones dentro de paréntesis siempre se evalúan primero. Permiten anular la precedencia por defecto.
2.  **Operadores Unarios**: Operadores que actúan sobre un solo operando, como el signo negativo (`-`), el operador de negación lógica (`not`), etc.
3.  **Potencia `**`** (exponenciación).
4.  **Multiplicación `*`, División `/`, División entera `//`, Módulo `%`**.
5.  **Suma `+`, Resta `-`**.
6.  **Operadores de Desplazamiento (Bitwise Shift)**: `<<`, `>>`.
7.  **Operadores Relacionales / de Comparación**: `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `is not`, `in`, `not in`.
8.  **Operadores Lógicos (Bitwise AND, OR, XOR)**: `&`, `^`, `|`.
9.  **Operadores Lógicos Booleanos**: `not` (ya mencionado como unario), `and`, `or`.
10. **Operadores de Asignación**: `=`, `+=`, `-=`, `*=`, `/=`, etc.

#### Ejemplo en Python:

Veamos un ejemplo simple para ilustrar cómo funciona la precedencia en Python:

````python
resultado1 = 10 + 5 * 2 # Sin paréntesis
print(f"Resultado 1 (10 + 5 * 2): {resultado1}") # Se evalúa 5 * 2 primero, luego la suma

resultado2 = (10 + 5) * 2 # Con paréntesis
print(f"Resultado 2 ((10 + 5) * 2): {resultado2}") # Se evalúa (10 + 5) primero, luego la multiplicación

resultado3 = 2 ** 3 + 4 * 2
print(f"Resultado 3 (2 ** 3 + 4 * 2): {resultado3}") # 2**3 (8), luego 4*2 (8), luego 8+8

resultado4 = not True and False or True
print(f"Resultado 4 (not True and False or True): {resultado4}") # not True (False), luego False and False (False), luego False or True (True)
````

### El Operador de Módulo (`%`)

El operador de módulo, representado por el símbolo `%`, es un operador aritmético que devuelve el **resto** de una división. En otras palabras, si divides un número `a` entre un número `b` (es decir, `a / b`), el resultado del operador de módulo `a % b` es lo que queda después de realizar la división entera.

**Sintaxis básica:**

```python
resultado = dividendo % divisor
```

Por ejemplo:
*   `10 % 3` es `1` (porque 10 dividido por 3 es 3 con un resto de 1).
*   `10 % 2` es `0` (porque 10 dividido por 2 es 5 con un resto de 0).
*   `7 % 5` es `2` (porque 7 dividido por 5 es 1 con un resto de 2).

#### Casos de Uso Prácticos:

El operador de módulo es increíblemente versátil y se utiliza en muchas situaciones, incluyendo:

1.  **Comprobar si un número es par o impar**: Si un número dividido por 2 tiene un resto de 0, es par. Si el resto es 1, es impar.
2.  **Determinar si un número es divisible por otro**: Si `a % b` es 0, significa que `a` es divisible por `b`.
3.  **Operaciones cíclicas o rotaciones**: Útil para mantener un valor dentro de un rango específico, como los días de la semana, las horas en un reloj o índices en una lista que se repiten.
4.  **Extraer el último dígito de un número**: El módulo 10 (`% 10`) de un número entero siempre dará su último dígito.
5.  **Conversión de unidades de tiempo**: Por ejemplo, calcular los minutos restantes después de convertir segundos a horas.


````python
# 1. Comprobar si un número es par o impar
numero1 = 10
numero2 = 7

print(f"¿{numero1} es par? {numero1 % 2 == 0}")
print(f"¿{numero2} es par? {numero2 % 2 == 0}")

# 2. Determinar si un número es divisible por otro
num = 15
divisor = 3
print(f"¿{num} es divisible por {divisor}? {num % divisor == 0}")

num = 17
divisor = 3
print(f"¿{num} es divisible por {divisor}? {num % divisor == 0}")

# 3. Operaciones cíclicas (ej. días de la semana)
dias_semana = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
dia_actual = 0 # Lunes

# ¿Qué día será en 3 días?
proximo_dia_indice = (dia_actual + 3) % 7
print(f"Si hoy es {dias_semana[dia_actual]}, en 3 días será {dias_semana[proximo_dia_indice]}")

# ¿Qué día será en 10 días?
proximo_dia_indice_futuro = (dia_actual + 10) % 7
print(f"Si hoy es {dias_semana[dia_actual]}, en 10 días será {dias_semana[proximo_dia_indice_futuro]}")

# 4. Extraer el último dígito de un número
numero_largo = 1234567
ultimo_digito = numero_largo % 10
print(f"El último dígito de {numero_largo} es: {ultimo_digito}")

# 5. Conversión de unidades de tiempo (ej. segundos a minutos y segundos)
total_segundos = 150
minutos = total_segundos // 60 # División entera para obtener los minutos
segundos_restantes = total_segundos % 60 # Módulo para obtener los segundos restantes
print(f"{total_segundos} segundos son {minutos} minutos y {segundos_restantes} segundos.")
````
