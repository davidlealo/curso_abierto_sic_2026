**input()** lee la entrada del usuario y la guarda en una variable de tipo string

```python
numero_string = input()
```

**type()** nos dice el tipo de dato al que corresponde la variable

```python
print(type(numero_string))
```

**int()** transforma una variable a tipo entero

```python
numero_int = int(numero_string)
```

```python
print(type(numero_int))
```

**bin()** transforma una variable a tipo binario

```python
numero_binario = bin(numero_int)
print(numero_binario)
```
