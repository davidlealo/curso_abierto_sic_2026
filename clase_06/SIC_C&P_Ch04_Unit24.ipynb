### 3. Solution

```python
def do_experiment(floor, breaking):
    return floor >= breaking

def find_highest_safe_floor(height, breaking):
    for n in range(1, height + 1):
        if do_experiment(n, breaking):
            return n - 1
    return height
```

```python
from random import randint

height = int(input("Input the number of floors: "))
breaking = randint(1, height)
floor = find_highest_safe_floor(height, breaking)
print(f"Your egg will safe till the {floor}-th floor.")
```

### 2. Sequential Search

```python
def seq_search (nums, x):
    for i in range(len(nums)):
        if x == nums[i]:
            return i
    return -1
```

```python
S = [11, 37, 45, 26, 59, 28, 17, 53]
x = 53
pos = seq_search (S, x)
print (f"In S, {x} is at position {pos}.")
```

### 3. Finding the Largest

```python
def find_largest(nums):
    largest = 0
    for i in range(1, len(nums)):
        if nums[largest] < nums[i]:
            largest = i
    return largest
```

```python
nums = [11, 37, 45, 26, 59, 28, 17, 53]
pos = find_largest(nums)
print(f"The largest is {nums[pos]} at {pos}")
```

```python
def find_two (nums):
    x = y = 0
    for i in range(1, len(nums)):
        if nums[x] < nums[i]:
            x = i
        elif nums[y] > nums[i]:
            y = i
    return x, y
```

```python
nums = [11, 37, 45, 26, 59, 28, 17, 53]
i, j = find_two(nums)
print(nums[i], nums [j])
```
