## 1. Funciones de Tipo y Conversión

### `int(x)`
Convierte un valor a número entero.
```python
int("42")       # → 42
int(3.99)       # → 3
int("1010", 2)  # → 10 (binario a decimal)
```

### `float(x)`
Convierte un valor a número de punto flotante.
```python
float("3.14")  # → 3.14
float(5)       # → 5.0
```

### `str(x)`
Convierte un valor a cadena de texto.
```python
str(100)    # → "100"
str(True)   # → "True"
str([1, 2]) # → "[1, 2]"
```

### `bool(x)`
Convierte un valor a booleano.
```python
bool(0)    # → False
bool("")   # → False
bool("hi") # → True
bool(42)   # → True
```

### `list(iterable)`
Crea una lista a partir de un iterable.
```python
list("abc")      # → ['a', 'b', 'c']
list(range(5))   # → [0, 1, 2, 3, 4]
list((1, 2, 3))  # → [1, 2, 3]
```

### `tuple(iterable)`
Crea una tupla a partir de un iterable.
```python
tuple([1, 2, 3])  # → (1, 2, 3)
tuple("abc")      # → ('a', 'b', 'c')
```

### `set(iterable)`
Crea un conjunto (sin duplicados) a partir de un iterable.
```python
set([1, 2, 2, 3, 3])  # → {1, 2, 3}
set("hello")           # → {'h', 'e', 'l', 'o'}
```

### `dict(**kwargs)` / `dict(iterable)`
Crea un diccionario.
```python
dict(nombre="Ana", edad=30)       # → {'nombre': 'Ana', 'edad': 30}
dict([("a", 1), ("b", 2)])        # → {'a': 1, 'b': 2}
```

### `frozenset(iterable)`
Crea un conjunto inmutable.
```python
fs = frozenset([1, 2, 3])
# fs.add(4)  ← Error: no se puede modificar
```

### `bytes(x)` / `bytearray(x)`
Crea objetos de bytes.
```python
bytes(5)           # → b'\x00\x00\x00\x00\x00'
bytearray(b"abc")  # → bytearray(b'abc')
```

### `complex(real, imag)`
Crea un número complejo.
```python
complex(2, 3)   # → (2+3j)
complex("3+4j") # → (3+4j)
```

## 2. Funciones Matemáticas

### `abs(x)`
Retorna el valor absoluto.
```python
abs(-7)    # → 7
abs(-3.14) # → 3.14
```

### `round(x, ndigits)`
Redondea un número.
```python
round(3.14159, 2)  # → 3.14
round(2.5)         # → 2  (redondeo bancario)
```

### `max(iterable)` / `max(a, b, ...)`
Retorna el valor máximo.
```python
max([3, 1, 4, 1, 5])   # → 5
max(10, 20, 30)         # → 30
max("abc", "xyz")       # → 'xyz'
```

### `min(iterable)` / `min(a, b, ...)`
Retorna el valor mínimo.
```python
min([3, 1, 4, 1, 5])  # → 1
min(10, 20, 30)        # → 10
```

### `sum(iterable, start=0)`
Suma los elementos de un iterable.
```python
sum([1, 2, 3, 4])     # → 10
sum([1, 2, 3], 100)   # → 106
```

### `pow(base, exp, mod=None)`
Eleva a una potencia (con módulo opcional).
```python
pow(2, 10)      # → 1024
pow(2, 10, 100) # → 24  (1024 % 100)
```

### `divmod(a, b)`
Retorna cociente y resto como tupla.
```python
divmod(17, 5)  # → (3, 2)  → 17 = 5*3 + 2
```

## 3. Funciones de Iteración y Secuencias

### `range(stop)` / `range(start, stop, step)`
Genera una secuencia de números.
```python
list(range(5))           # → [0, 1, 2, 3, 4]
list(range(2, 10, 2))    # → [2, 4, 6, 8]
list(range(10, 0, -2))   # → [10, 8, 6, 4, 2]
```

### `len(objeto)`
Retorna la longitud de un objeto.
```python
len([1, 2, 3])    # → 3
len("Python")     # → 6
len({"a": 1})     # → 1
```

### `enumerate(iterable, start=0)`
Agrega un contador a un iterable.
```python
for i, v in enumerate(["a", "b", "c"]):
    print(i, v)
# 0 a
# 1 b
# 2 c
```

### `zip(*iterables)`
Combina múltiples iterables elemento a elemento.
```python
nombres = ["Ana", "Luis", "Eva"]
edades  = [30, 25, 28]
list(zip(nombres, edades))
# → [('Ana', 30), ('Luis', 25), ('Eva', 28)]
```

### `map(función, iterable)`
Aplica una función a cada elemento.
```python
list(map(str.upper, ["hola", "mundo"]))  # → ['HOLA', 'MUNDO']
list(map(lambda x: x**2, [1, 2, 3, 4])) # → [1, 4, 9, 16]
```

### `filter(función, iterable)`
Filtra elementos según una condición.
```python
pares = list(filter(lambda x: x % 2 == 0, range(10)))
# → [0, 2, 4, 6, 8]
```

### `sorted(iterable, key=None, reverse=False)`
Retorna una lista ordenada.
```python
sorted([3, 1, 4, 1, 5])                       # → [1, 1, 3, 4, 5]
sorted(["banana", "apple", "kiwi"], key=len)   # → ['kiwi', 'apple', 'banana']
sorted([3, 1, 4], reverse=True)                # → [4, 3, 1]
```

### `reversed(secuencia)`
Retorna un iterador invertido.
```python
list(reversed([1, 2, 3, 4]))  # → [4, 3, 2, 1]
```

### `any(iterable)`
Retorna `True` si al menos un elemento es verdadero.
```python
any([False, False, True])  # → True
any([0, 0, 0])             # → False
```

### `all(iterable)`
Retorna `True` si **todos** los elementos son verdaderos.
```python
all([True, True, True])   # → True
all([True, False, True])  # → False
```

## 4. Funciones de Entrada/Salida

### `print(*objetos, sep=' ', end='\n', file=sys.stdout)`
Imprime en la salida estándar.
```python
print("Hola", "Mundo")              # → Hola Mundo
print("A", "B", "C", sep="-")      # → A-B-C
print("Sin salto", end="")
```

### `input(prompt="")`
Lee una línea desde la entrada estándar.
```python
nombre = input("¿Cuál es tu nombre? ")
edad   = int(input("¿Cuántos años tienes? "))
```

### `open(archivo, modo='r')`
Abre un archivo y retorna un objeto de archivo.
```python
# Lectura
with open("datos.txt", "r", encoding="utf-8") as f:
    contenido = f.read()

# Escritura
with open("salida.txt", "w", encoding="utf-8") as f:
    f.write("Hola archivo\n")
```

| Modo | Descripción |
|------|-------------|
| `r`  | Lectura (por defecto) |
| `w`  | Escritura (sobrescribe) |
| `a`  | Agregar al final |
| `rb` | Lectura binaria |
| `wb` | Escritura binaria |

## 5. Funciones de Objetos y Atributos

### `type(objeto)`
Retorna el tipo de un objeto.
```python
type(42)       # → <class 'int'>
type("hola")   # → <class 'str'>
type([1, 2])   # → <class 'list'>
```

### `isinstance(objeto, clase)`
Verifica si un objeto es instancia de una clase.
```python
isinstance(42, int)         # → True
isinstance("hola", (int, str))  # → True
```

### `issubclass(subclase, clase)`
Verifica si una clase es subclase de otra.
```python
issubclass(bool, int)   # → True
issubclass(int, float)  # → False
```

### `hasattr(objeto, nombre)`
Verifica si un objeto tiene un atributo.
```python
hasattr([1, 2, 3], "append")  # → True
hasattr(42, "upper")           # → False
```

### `getattr(objeto, nombre, default)`
Obtiene el valor de un atributo.
```python
class Persona:
    nombre = "Ana"

p = Persona()
getattr(p, "nombre")           # → 'Ana'
getattr(p, "edad", "N/A")      # → 'N/A'
```

### `setattr(objeto, nombre, valor)`
Establece el valor de un atributo.
```python
setattr(p, "edad", 30)
p.edad  # → 30
```

### `delattr(objeto, nombre)`
Elimina un atributo de un objeto.
```python
delattr(p, "edad")
```

### `dir(objeto)`
Retorna lista de atributos y métodos del objeto.
```python
dir([])       # → ['append', 'clear', 'copy', ...]
dir(str)      # → lista de métodos de str
```

### `vars(objeto)`
Retorna el diccionario `__dict__` del objeto.
```python
vars(p)  # → {'nombre': 'Ana'}
```

### `id(objeto)`
Retorna el identificador único del objeto en memoria.
```python
a = [1, 2, 3]
id(a)  # → 140234567890 (varía por ejecución)
```

### `hash(objeto)`
Retorna el hash de un objeto (si es hashable).
```python
hash("python")  # → algún entero
hash(42)        # → 42
```

## 6. Funciones de Clases y Herencia

### `super()`
Retorna un proxy del objeto padre.
```python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

class Perro(Animal):
    def __init__(self, nombre, raza):
        super().__init__(nombre)
        self.raza = raza
```

### `classmethod` / `staticmethod`
Decoradores para métodos de clase y estáticos.
```python
class Circulo:
    PI = 3.14159

    @classmethod
    def desde_diametro(cls, d):
        return cls(d / 2)

    @staticmethod
    def es_radio_valido(r):
        return r > 0
```

### `property`
Crea atributos gestionados con getters/setters.
```python
class Temperatura:
    def __init__(self):
        self._celsius = 0

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, valor):
        if valor < -273.15:
            raise ValueError("Temperatura inválida")
        self._celsius = valor
```

## 7. Funciones de Alcance y Variables

### `globals()`
Retorna el diccionario de variables globales.
```python
x = 10
globals()["x"]  # → 10
```

### `locals()`
Retorna el diccionario de variables locales.
```python
def ejemplo():
    a, b = 1, 2
    print(locals())  # → {'a': 1, 'b': 2}
```

### `callable(objeto)`
Verifica si un objeto es llamable.
```python
callable(print)     # → True
callable(42)        # → False
callable(lambda: 0) # → True
```

## 8. Funciones de Evaluación y Ejecución

### `eval(expresión)`
Evalúa una expresión Python como cadena.
```python
eval("2 + 2")           # → 4
eval("len([1, 2, 3])")  # → 3
```
> ⚠️ **Precaución**: Nunca uses `eval()` con entradas no confiables.

### `exec(código)`
Ejecuta código Python como cadena.
```python
exec("x = 10\nprint(x)")  # → 10
```
> ⚠️ **Precaución**: Mismo riesgo de seguridad que `eval()`.

### `compile(fuente, nombre, modo)`
Compila código fuente a bytecode.
```python
codigo = compile("print('Hola')", "<string>", "exec")
exec(codigo)  # → Hola
```

## 9. Otras Funciones Útiles

### `repr(objeto)`
Retorna representación oficial del objeto.
```python
repr("hola")    # → "'hola'"
repr([1, 2, 3]) # → '[1, 2, 3]'
```

### `chr(código)` / `ord(carácter)`
Convierte entre carácter y código Unicode.
```python
chr(65)   # → 'A'
chr(9829) # → '♥'
ord('A')  # → 65
ord('♥')  # → 9829
```

### `bin(n)` / `oct(n)` / `hex(n)`
Convierte enteros a representación binaria, octal o hexadecimal.
```python
bin(255)  # → '0b11111111'
oct(255)  # → '0o377'
hex(255)  # → '0xff'
```

### `format(valor, formato)`
Formatea un valor según una especificación.
```python
format(3.14159, ".2f")   # → '3.14'
format(255, "08b")        # → '11111111'
format(1000000, ",")      # → '1,000,000'
```

### `slice(start, stop, step)`
Crea un objeto slice reutilizable.
```python
s = slice(1, 5, 2)
[0, 1, 2, 3, 4, 5][s]  # → [1, 3]
```

### `iter(objeto)` / `next(iterador)`
Trabaja con iteradores manualmente.
```python
it = iter([10, 20, 30])
next(it)  # → 10
next(it)  # → 20
next(it)  # → 30
next(it, "fin")  # → 'fin'  (valor por defecto)
```

### `zip_longest` → usar `itertools`
> Para un zip que no se detenga en el más corto, usa `itertools.zip_longest`.

### `memoryview(objeto_bytes)`
Crea una vista de memoria de un objeto de bytes.
```python
datos = bytearray(b"Python")
mv = memoryview(datos)
mv[0]  # → 80  (código ASCII de 'P')
```

### `object()`
Retorna un nuevo objeto base (la clase raíz de todo en Python).
```python
o = object()
type(o)  # → <class 'object'>
```
